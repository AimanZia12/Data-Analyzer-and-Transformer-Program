```python
print("\n------Welcome to the Data Analyzer and Transformer Program------")
while True:
    print("\n Main Menu")
    print("1.Input Data")
    print("2.Display Data Summary (Built-in Functions")
    print("3.Calculate Factorial (Recursion)")
    print("4.Filter Data by Threshold (Lambda Function)")
    print("5.Display Dataset Statistics(Return Multiple Values)")
    print("6.Sort Data")
    print("7.Exit Program")

    user_choice = input("Please enter your choice:")

    if user_choice == "1":
        print("\n----Please enter your data----")
        
        Data_Mode = input("Inserting mode for data (MANUAL/ SAMPLE): ").upper()
        if Data_Mode == "MANUAL":
            Data = list(map(int, input("Enter numbers separated by space: ").split()))
        else:
            Data = [10, 60, 20, 30, 40, 50, 20]
        print(Data)
        
    elif user_choice == "2":
        print("\n-----Display Data Summary-----")
        
        def length(Data):
            """ This function is used to calculate total elements of data."""
            return len(Data)
        d = length(Data)
        print("Total Elements:", d)

        def Sum(Data):
            """This function is used to calculate sum of data."""
            return sum(Data)
        d = Sum(Data)
        print("Sum of all values:", d)

        def Maximum(Data):
            """This function is used to find maximum value in data."""
            return max(Data)
        d = Maximum(Data)
        print("Maximum Value:", d)

        def Minimum(Data):
            """This function is used to find minimum value in data."""
            return min(Data)
        d = Minimum(Data)
        print("Minimum Value:", d)
        
        def average(Data):
            """This function is used to find average of data."""
            return sum(Data)/ len(Data)
        d = average(Data)
        print("Average of data :", d)

        def duplicate(Data):
            """This function is used to find duplicate values in data."""
            duplicates = []
            for item in Data:
                if Data.count(item) > 1:
                    duplicates.append(item)
                    
            if duplicates:
                return f"It has {len(duplicates)} duplicates and duplicates are: {duplicates}"
                
            return "No Duplicate"
       
        c = duplicate(Data)
        print("Duplicate Data Status:", c)
        
        def Unique(Data):
            """ This function is used to find unique values in data."""
            Unique_items = []
            for item in Data:
                if Data.count(item) == 1:
                    Unique_items.append(item)
                    
            if Unique_items:
                return f"It has {len(Unique_items)} unique items and unique items are: {Unique_items}"
                
            return "No Unique Item"
       
        c = Unique(Data)
        print("Unique Data Status:", c)
        

    elif user_choice == "3":
        print("\n-------Calculate Factorial-----")

        n = int(input("Enter a number for which you want to find factorial:"))
        
        def factorial(n):
            """ This function is used to find factorial of numbers."""
            if (n == 0 or n == 1):  
                return 1
            else:
                return n * factorial(n - 1) 
        print(factorial(n))
        

    elif user_choice == "4":
        print("\n------Filter Data by Threshold------")

        t = int(input("Enter a threshold value for filtering data"))
        greater_values = list(filter(lambda x: x > t , Data))
        print(greater_values)

        cube_of_values = list(map(lambda x: x % 10 == 0, Data))
        print(cube_of_values)

    elif user_choice == "5":
        print("\n------Display Dataset Statistics------")

        Summary_of_Dataset = {}
        print("\n------User gives multiple values------")
        def multiple_values(*args):
            """Using *args this function accepts multiple values."""
            global Summary_of_Dataset
            
            for value in args:
                print(value, end=" ")
            print()
            Summary_of_Dataset = {
                    "Total_value": len(args),
                    "Sum_of_value": sum(args),
                    "Maximum_of_value": max(args),
                    "Minimum_of_value": min(args),
                    "Average_of_value": sum(args) / len(args)
                }
        def summary(**kwargs):
            """Using **kwargs this function displays summary."""
            print("\n------ DATASET SUMMARY -----")
            for key, value in kwargs.items():
                print(f"{key}: {value}")
        multiple_values(10, 80, 20, 30, 40, 50, 20)
        summary(**Summary_of_Dataset)   

    elif user_choice == "6":
        print("\n-----Sort Data-----")
        Data_Mode = input("Inserting mode for data (MANUAL/ SAMPLE): ").upper()
        if Data_Mode == "MANUAL":
            Data = list(map(int, input("Enter numbers separated by space: ").split()))
        else:
            Data = [10, 20, 30, 40, 50, 20]

            Original_Data = Data.copy()
        print("\n-----Choose Sorting methods----")
        print("1.Ascending Order")
        print("2.Descending Order")
        
        method = input("Enter your choice:")
        if method == "1":
            
            Data.sort()
            New_Data = sorted(Data)
            print("Data in Ascending Order is :", Data)
            print("New Sorted Data is :", New_Data)
            print("Original Data is:", Original_Data)
            
        elif method == "2":
            Data.sort(reverse = True)
            New_Data = sorted(Data, reverse = True)
            print("Data in Descending Order is :", Data)
            print("New Sorted Data is :", New_Data)
            print("Original Data is:", Original_Data)

    elif user_choice == "7":
        print("\n ----Exit the Program---")
        print("\nThank you, for choosing the Data Analyzer and Transformer Program")
        break
    else:
        print("\n Invalid")
        
        
```
